```python
from pathlib import Path
import json
from pprint import pprint

# 1) JSON 파일이 있는 디렉터리 경로를 지정하세요
directory = Path("/data/user_data/mineuih/womd/training")

# 2) 디렉터리 안의 .json 파일 목록 수집
json_files = sorted(directory.glob("*.json"))
json_files = [f for f in json_files if "store" not in f.name]  # "store"가 파일명에 포함된 경우 제외
if not json_files:
    raise FileNotFoundError(f"{directory} 에 .json 파일이 없습니다.")

# 3) 불러올 파일 선택 (예: 첫 번째 파일)
target_file = json_files[0]

print(f"선택된 파일: {target_file}")

# 4) JSON 로드 및 내용 출력
with target_file.open("r", encoding="utf-8") as f:
    data = json.load(f)

print("\nJSON 내용:")
pprint(data)
```
